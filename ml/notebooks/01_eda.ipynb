# Step 1 & 3A — Exploratory Data Analysis (EDA)
### Ransomware Early-Warning System for Small Institutions

This notebook analyzes class distributions, behavioral feature distributions, correlation structures, and feature skewness on the training data strictly.

```python
import os
import sys
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

project_root = os.path.abspath('..')
if project_root not in sys.path:
    sys.path.insert(0, project_root)

X_train = pd.read_csv(os.path.join(project_root, 'data', 'processed', 'X_train.csv'))
y_train = pd.read_csv(os.path.join(project_root, 'data', 'processed', 'y_train.csv'))['target'].values
with open(os.path.join(project_root, 'data', 'raw', 'MLRan', 'RFE_selected_feature_names_dic.json'), 'r') as f:
    rfe_dict = json.load(f)

print(f'X_train shape: {X_train.shape}')
print(f'y_train count: {len(y_train)}')
```

## 1. Class Distribution Analysis
Verifies goodware vs. ransomware balance in the training partition.

```python
class_counts = pd.Series(y_train).value_counts()
print('Class 0 (Goodware):', class_counts.get(0))
print('Class 1 (Ransomware):', class_counts.get(1))

plt.figure(figsize=(6, 4))
sns.barplot(x=['Goodware (0)', 'Ransomware (1)'], y=[class_counts[0], class_counts[1]], palette=['#2b5c8f', '#d95f02'])
plt.title('Class Distribution in X_train')
plt.ylabel('Count')
plt.show()
```

## 2. Behavioral Categories Breakdown
Categorizes the 483 behavioral features by system domain (API, Registry, File, Process, etc.).

```python
categories = {}
for col in X_train.columns:
    name = rfe_dict.get(col, 'OTHER')
    cat = name.split(':')[0] if ':' in name else 'OTHER'
    categories[cat] = categories.get(cat, 0) + 1

cat_df = pd.DataFrame(list(categories.items()), columns=['Category', 'Count']).sort_values('Count', ascending=False)
plt.figure(figsize=(10, 4))
sns.barplot(data=cat_df, x='Count', y='Category', palette='mako')
plt.title('Features by Behavioral Category')
plt.show()
```
